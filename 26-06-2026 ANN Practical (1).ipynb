```python
# Import the necessary liabraries
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow import keras
from sklearn.model_selection import train_test_split
from tensorflow.keras import layers
```

```python
# Activation functions (Binary sigmoid)
# Layers
# Forwar & Backword Propogation
# Loss Function
# Optimizer
# Epochs
```

```python

np.random.seed(42)

# Number of rows for this
n = 500

# Input features

annual_income = np.random.randint(20000, 200000, n)
credit_score = np.random.randint(300, 850, n)
loan_amount = np.random.randint(5000, 100000, n)
employment_years = np.random.randint(0, 30, n)

# Output Feature

loan_approved = (
    (annual_income > 50000) &
     (credit_score > 600 ) &
    (loan_amount < annual_income * 0.8)
).astype(int)
```

```python
df = pd.DataFrame({
    'annual_income': annual_income,
    'credit_score': credit_score,
    'loan_amount': loan_amount,
    'employment_years': employment_years,
    'loan_approved': loan_approved
})
```

```python
X = df[['annual_income', 'credit_score', 'loan_amount', 'employment_years']]
y = df['loan_approved']
```

```python
X.head(3)
```

```python
# Scaling
# Normalization
# Standardization
X_min = X.min()
X_max = X.max()
X_scaled = (X - X_min) / (X_max - X_min)
```

```python
X_scaled
```

```python
# Training - Testing Spliting
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)
```

```python
# Model Creating
model = keras.Sequential([
    layers.Input(shape = (X_train.shape[1],)), # Input Layer
    layers.Dense(8,activation='relu'), # Hidden Layer
    layers.Dense(1,activation='sigmoid') # Output Layer
])
```

```python
model.compile(optimizer='sgd',loss='binary_crossentropy',metrics=['accuracy'])
```

```python
history = model.fit(X_train,y_train, validation_data=(X_test,y_test),
                    epochs = 100, batch_size = 4, verbose=1)
```
